```python
# Set Seaborn style
sns.set(style='whitegrid')
```

```python
# Q1 What genres are most popular amoung bestsellers ?
plt.figure(figsize=(6,4))
sns.countplot(x='Genre', data=df, palette='tab10')
plt.title('Fiction vs Non-Fiction Bestsellers')
plt.xlabel('Genre')
plt.ylabel('Count of Books')
plt.show()
```

#### observations :- non fiction is most popular amoung bestseller.

```python
# Q2 Which authors have multiple bestselling books?
top_authors = df['Author'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_authors.index, y=top_authors.values, palette='tab10')
plt.title('Top 10 Authors with Multiple Books')
plt.xlabel('Author')
plt.ylabel('Number of Books')
plt.xticks(rotation=45)
plt.show()
```

#### Observations :- jeff kinney , rick riordan and stephenie meyer  has top 3 authors have mutiple bestselling books

```python
# Q3 How has the number of bestselling books changed each year from 2009–2019?
yearly_books = df.groupby('Year')['Name'].count()
plt.figure(figsize=(10,6))
sns.lineplot(x=yearly_books.index, y=yearly_books.values, marker='o',palette='tab10')
plt.title('Number of Bestselling Books Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Books')
plt.show()
```

#### Observation :- after 2010 the sell was decresing and after 2012 to 2018 the sell of books was almost constant.higest sell year is 2009 around 50 books was sold thats year

```python
# Q4. What is the typical price range for bestselling books?
plt.figure(figsize=(8,6))
sns.histplot(df['Price'], bins=30, kde=True, color='green')
plt.title('Distribution of Book Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

#### Observation :- the typical price range for best selling book is 2 to 20 $
#### also above historam show also right screwed data so if a buid model need to be scalled

```python
# Q5 Is there a relationship between price and user rating?
corr = df[['Price', 'User Rating']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation between Price and User Rating')
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(10,6))
sns.scatterplot(x='Price', y='User Rating', data=df, hue='Genre', palette='tab10')
plt.title('Relationship between Price and User Rating')
plt.xlabel('Price')
plt.ylabel('User Rating')
plt.show()
```

```python
#user raating distribution
plt.figure(figsize=(8,6))
sns.histplot(df['User Rating'], bins=20, kde=True, color='green')
plt.title('Distribution of User Ratings')
plt.xlabel('User Rating')
plt.ylabel('Frequency')
plt.show()
```

#### observation :- there no corretion between user rating and price thats proof with heat map and scatter plot . Also there are more insights user rating above majority 4.0 above thats very good sign thats most of books higly liked and popular amoung all users

```python
# Q6. Are Fiction books more highly rated than Non-Fiction books?
compare = df.groupby('Genre')['User Rating'].mean()
plt.figure(figsize=(8,6))
sns.barplot(x=compare.index, y=compare.values, palette='tab10')
plt.title('Average User Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average User Rating')
plt.show()
```

#### Observation :- no both are qual rating distibution fiction and non fiction

```python
# Q7 Which years had the highest or lowest numbers of bestsellers?
higest_sellar = df.groupby('Year')['Name'].count()
plt.figure(figsize=(10,6))
sns.barplot(x=higest_sellar.index, y=higest_sellar.values, palette='tab10')
plt.title('Number of Bestselling Books Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Books')
plt.show()
```

#### Observation :- highest sales in 2009 which is 50 and lowest sell in 2014 which is 26

```python
# extra question . Top 10 Books by Estimated Revenue
top_books = df.nlargest(10, 'Estimated Revenue')
#bar chart
plt.figure(figsize=(10,6))
sns.barplot(x='Estimated Revenue', y='Name', data=top_books, palette='tab10')
plt.title('Top 10 Books by Estimated Revenue')
plt.xlabel('Book Name')
plt.ylabel('Estimated Revenue')
plt.xticks(rotation=90)
plt.show()
```

```python
# Q8 What are the trends in user ratings over the years?
avg_rating_year = df.groupby('Year')['User Rating'].mean()
plt.figure(figsize=(10,6))
sns.lineplot(x=avg_rating_year.index, y=avg_rating_year.values, marker='o', palette='tab10')
plt.title('Average User Rating Over the Years')
plt.xlabel('Year')
plt.ylabel('Average User Rating')
plt.show()
```

#### Observation :- after 2012 user rating continuoly incresing over the year

```python
# Q9 What are the top 10 highest-rated books?
top = df.nlargest(10, 'User Rating')
#bar chart
plt.figure(figsize=(10,6))
sns.barplot(x='User Rating', y='Name', data=top, palette='tab10')
plt.title('Top 10 Highest-Rated Books')
plt.xlabel('User Rating')
plt.ylabel('Book Name')
plt.xticks(rotation=90)
plt.show()
```

#### Observation :- above all top 10 books name was higest rating around 4.9

```python
# Q10 . Are there any particular pricing strategies successful authors follow?
# Top 5 Authors with Most Books
top_authors_list = df['Author'].value_counts().head(5).index

# Filter dataset
top_authors_books = df[df['Author'].isin(top_authors_list)]

# Plot
plt.figure(figsize=(10,6))
sns.boxplot(x='Author', y='Price', data=top_authors_books, palette='Set3')
plt.title('Price Distribution among Top 5 Authors')
plt.ylabel('Price (USD)')
plt.xlabel('Author')
plt.xticks(rotation=45)
plt.show()
```

```python
#most sold books per aothers
sold_books_count= df.groupby('Author')['Name'].count().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=sold_books_count.index, y=sold_books_count.values, palette='tab10')
plt.title('Top 10 Authors with Most Books')
plt.xlabel('Author')
plt.ylabel('Number of Books')
plt.xticks(rotation=45)
plt.show()
```

#### Observations :- Author | Price Strategy Insight
1. Stephenie Meyer | Wide price range (0 to 80+). One very expensive book (outlier). Otherwise, normal books around $10–$20.

2. Jeff Kinney | Mostly affordable books ($5–$15), few slightly premium books up to $22.

3. Dav Pilkey | Consistent low pricing ($5–$10). Very stable, no expensive books.

4. Bill O'Reilly | Medium pricing ($5–$12) but one big outlier (~$25).

5. Rick Riordan | Mostly between $8–$14. Small variation, safe pricing strategy.


#### conclusion:- Stephenie Meyer sometimes prices luxury (premium) editions ,Dav Pilkey and Jeff Kinney focus on affordable, consistent pricing (good for children's books market),Bill O'Reilly and Rick Riordan keep their prices moderate and stable, ideal for mass-market.

## Analyze Amazon’s Top 50 Bestselling Books (2009–2019) to understand patterns in book sales, pricing, and customer preferences.


##  Key Findings:

1. **Fiction dominates** slightly over Non-Fiction in the bestseller list.

##  Actionable Recommendations:

-  **Pricing Strategy:** Keep book pricing mostly within the $5–$15 range to boost accessibility and volume.

-  **Prioritize Ratings:** Maintain high-quality content; encourage verified reviews to sustain high ratings.
