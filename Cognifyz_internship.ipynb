### task 1
---Task: Restaurant Reviews

Analyze the text reviews to identify the most
common positive and negative keywords.

Calculate the average length of reviews and
explore if there is a relationship between
review length and rating.

```python
#Analyze the text reviews to identify the most common positive and negative keywords.
# defining function if excellen,very good,good,average put positive keyword other put negative keyworld
def classify_review(review):
    if review in ['Excellent', 'Very Good', 'Good']:
        return 'Positive'
    elif review == 'Average':
        return 'Average'
    else:
        return 'Negative'

# count all positive and negative def function
def count_positive_negative(reviews):
    positive_count = sum(1 for review in reviews if review == 'Positive')
    negative_count = sum(1 for review in reviews if review == 'Negative')
    return positive_count, negative_count

#use case
df['Review Classification'] = df['Rating text'].apply(classify_review)
positive_count, negative_count = count_positive_negative(df['Review Classification'])
```
