```python
# Chart - 4 Multigenre content is rising

import matplotlib.pyplot as plt
# Chart visualization code
# Calculate genres_count on the titles_df before grouping
titles_df['genres_count'] = titles_df['genres'].apply(lambda x: len(x) if isinstance(x, list) else 0)

# Group by release year and calculate the mean number of genres per title
genre_diversity_over_time = titles_df.groupby('release_year')['genres_count'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='release_year', y='genres_count', data=genre_diversity_over_time)
plt.title('Average Number of Genres Per Title Over Time')
plt.xlabel('Release Year')
plt.ylabel('Average Number of Genres')
plt.grid(True)
plt.show()
```
